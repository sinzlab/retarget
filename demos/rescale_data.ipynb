```python
%load_ext autoreload
%autoreload 2

from retarget.utils.BVH import load, save
```

```python
scales = {
    "bandai-namco": 100,
    "cmu": 2.54 / 0.45,
    "sfu": 1,
    'pfnn': 1,
    'acaad': 1,
    'totalcapture': 1,
    'lafan1': 1,
    'mixamo': 1,
    'truebones': 1,
}
```

```python
from pathlib import Path
from tqdm import tqdm

bandai_namco_path = Path('../data/test_with_truebones/bandai-namco')
output_path = Path('../data/test_with_truebones/bandai-namco-scaled')

output_path.mkdir(parents=True, exist_ok=True)

bandai_namco_files = list(bandai_namco_path.glob('**/*.bvh'))

for file in tqdm(bandai_namco_files):
    animation, names, frametime = load(file)

    animation.positions *= scales['bandai-namco']
    animation.offsets *= scales['bandai-namco']

    save(output_path / file.name, animation, names=names, frametime=1/30)
```

```python
from pathlib import Path
from tqdm import tqdm

bandai_namco_path = Path('../data/train_with_truebones/cmu')
output_path = Path('../data/train_with_truebones/cmu-scaled')

output_path.mkdir(parents=True, exist_ok=True)

bandai_namco_files = list(bandai_namco_path.glob('**/*.bvh'))

for file in tqdm(bandai_namco_files):
    animation, names, frametime = load(file)

    animation.positions *= scales['cmu']
    animation.offsets *= scales['cmu']

    # downsample to 30 fps
    animation.rotations.qs = animation.rotations.qs[::4]
    animation.positions = animation.positions[::4]

    save(output_path / file.name, animation, names=names, frametime=1/30)
```
